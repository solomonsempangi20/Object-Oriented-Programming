# Boda Boda Fare Calculator

This calculator estimates a ride using a base fare and a charge for each kilometre.

```python
base_fare = 50
fare_per_kilometre = 15

distance = float(input("How many kilometres will you travel? "))

fare = base_fare + (distance * fare_per_kilometre)

print(f"Base fare: Ksh {base_fare}")
print(f"Distance fare: Ksh {distance * fare_per_kilometre:.2f}")
print(f"Total fare: Ksh {fare:.2f}")
```
