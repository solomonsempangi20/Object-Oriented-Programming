## The `Phone` class

```python
class Phone:
```

A class is a blueprint for creating objects. In this example, the `Phone` class describes what information and actions every phone should have.

## The constructor

```python
def __init__(self, brand, storage_capacity):
    self.brand = brand
    self.storage_capacity = storage_capacity
```

- `__init__` is the constructor. It runs automatically when a new `Phone` object is created.
- `brand` stores the phone manufacturer's name.
- `storage_capacity` stores the amount of storage in gigabytes.
- `self` refers to the current phone object.

## The methods

### `display_information()`

This method prints the phone's brand and storage capacity.

### `simulate_call(phone_number)`

This method accepts a phone number and prints a message showing that the phone is calling that number. It simulates a call but does not make a real call.

## Creating objects

```python
phone1 = Phone("Apple", 128)
phone2 = Phone("Samsung", 256)
```

These statements create two separate `Phone` objects:

- `phone1` is an Apple phone with 128 GB of storage.
- `phone2` is a Samsung phone with 256 GB of storage.

## Calling the methods

```python
phone1.display_information()
phone1.simulate_call("555-0101")
```

The dot operator (`.`) is used to call a method that belongs to an object. The same methods are then called for `phone2`.

````python
class Phone:
    def __init__(self, brand, storage_capacity):
        self.brand = brand
        self.storage_capacity = storage_capacity

    def display_information(self):
        print(f"Brand: {self.brand}")
        print(f"Storage capacity: {self.storage_capacity} GB")

    def simulate_call(self, phone_number):
        print(f"{self.brand} is calling {phone_number}...")


phone1 = Phone("iPhone 17", 128)
phone2 = Phone("Samsung Galaxy S23 Ultra", 256)

print("Phone 1")
phone1.display_information()
phone1.simulate_call("0784430877")

print("\nPhone 2")
phone2.display_information()
phone2.simulate_call("0788590809")
````
